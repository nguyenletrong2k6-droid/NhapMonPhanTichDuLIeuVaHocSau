```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("dulieuxettuyendaihoc.csv")
```

```python
print("--- 10 DÒNG ĐẦU TIÊN ---")
print(df.head(10))
```

```python
print("\n--- 10 DÒNG CUỐI CÙNG ---")
print(df.tail(10))
```

```python
tan_so_dt = df["DT"].value_counts(dropna=False)
tan_suat_dt = df["DT"].value_counts(dropna=False, normalize=True)

print("Bảng khảo sát cột DT trước khi xử lý:")
print(
    pd.DataFrame(
        {"Tần số": tan_so_dt, "Tần suất": tan_suat_dt.map("{:.2%}".format)}
    )
)

df["DT"] = df["DT"].fillna(0)

print(f"\nSố lượng giá trị thiếu còn lại ở cột DT: {df['DT'].isnull().sum()}")
```

```python
tan_so_t1 = df["T1"].value_counts(dropna=False)
tan_suat_t1 = df["T1"].value_counts(dropna=False, normalize=True)

print("Bảng khảo sát cột T1 trước khi xử lý:")
print(
    pd.DataFrame(
        {"Tần số": tan_so_t1, "Tần suất": tan_suat_t1.map("{:.2%}".format)}
    )
)

mean_t1 = df["T1"].mean()
df["T1"] = df["T1"].fillna(mean_t1)

print(
    f"\nGiá trị Mean của T1 dùng để điền: {mean_t1:.2f}"
)
print(f"Số lượng giá trị thiếu còn lại ở cột T1: {df['T1'].isnull().sum()}")
```

```python
cac_cot_diem = [
    "T1",
    "L1",
    "H1",
    "S1",
    "V1",
    "X1",
    "D1",
    "N1",
    "T2",
    "L2",
    "H2",
    "S2",
    "V2",
    "X2",
    "D2",
    "N2",
    "T6",
    "L6",
    "H6",
    "S6",
    "V6",
    "X6",
    "D6",
    "N6",
    "DH1",
    "DH2",
    "DH3",
]

for col in cac_cot_diem:
    if df[col].isnull().sum() > 0:
        mean_value = df[col].mean()
        df[col] = df[col].fillna(mean_value)

print("Đã hoàn thành xử lý dữ liệu thiếu bằng phương pháp Mean cho các cột điểm!")
```

```python
# Tính TBM1 cho lớp 10
df["TBM1"] = (
    df["T1"] * 2
    + df["L1"]
    + df["H1"]
    + df["S1"]
    + df["V1"] * 2
    + df["X1"]
    + df["D1"]
    + df["N1"]
) / 10

# Tính TBM2 cho lớp 11
df["TBM2"] = (
    df["T2"] * 2
    + df["L2"]
    + df["H2"]
    + df["S2"]
    + df["V2"] * 2
    + df["X2"]
    + df["D2"]
    + df["N2"]
) / 10

# Tính TBM3 cho lớp 12
df["TBM3"] = (
    df["T6"] * 2
    + df["L6"]
    + df["H6"]
    + df["S6"]
    + df["V6"] * 2
    + df["X6"]
    + df["D6"]
    + df["N6"]
) / 10

# In thử 5 dòng đầu của các cột mới tạo để kiểm tra
print(df[["TBM1", "TBM2", "TBM3"]].head())
```

```python
# Định nghĩa hàm logic xếp loại theo yêu cầu đề bài
def tinh_xep_loai(tbm):
    if tbm < 5.0:
        return "Y"
    elif 5.0 <= tbm < 6.5:
        return "TB"
    elif 6.5 <= tbm < 8.0:
        return "K"
    elif 8.0 <= tbm < 9.0:
        return "G"
    else:  # tbm >= 9.0
        return "XS"


# Áp dụng hàm vừa tạo vào các cột TBM bằng hàm .apply()
df["XL1"] = df["TBM1"].apply(tinh_xep_loai)
df["XL2"] = df["TBM2"].apply(tinh_xep_loai)
df["XL3"] = df["TBM3"].apply(tinh_xep_loai)

# In thử kết quả 5 dòng đầu
print(df[["TBM1", "XL1", "TBM2", "XL2", "TBM3", "XL3"]].head())
```

```python
# Áp dụng công thức chuẩn hóa Min-Max tương đương quy đổi hệ số điểm
df["US_TBM1"] = (df["TBM1"] / 10) * 4
df["US_TBM2"] = (df["TBM2"] / 10) * 4
df["US_TBM3"] = (df["TBM3"] / 10) * 4

# Xem thử kết quả quy đổi
print(df[["TBM1", "US_TBM1", "TBM2", "US_TBM2", "TBM3", "US_TBM3"]].head())
```

```python
# Định nghĩa hàm tính toán dựa trên điều kiện của từng dòng dữ liệu (axis=1)
def xet_tuyen(row):
    khoi = row["KT"]
    dh1, dh2, dh3 = row["DH1"], row["DH2"], row["DH3"]

    # Khối A và A1
    if khoi in ["A", "A1"]:
        diem_xt = (dh1 * 2 + dh2 + dh3) / 4
    # Khối B
    elif khoi == "B":
        diem_xt = (dh1 + dh2 * 2 + dh3) / 4
    # Các khối còn lại (C, D,...)
    else:
        diem_xt = (dh1 + dh2 + dh3) / 3

    # Đậu trả về 1, rớt trả về 0
    return 1 if diem_xt >= 5.0 else 0


# Áp dụng hàm xet_tuyen cho toàn bộ bảng theo chiều ngang từng dòng (axis=1)
df["KQXT"] = df.apply(xet_tuyen, axis=1)

# Thống kê xem có bao nhiêu bạn Đậu (1) và Rớt (0)
print(df["KQXT"].value_counts())
```

```python
# Xuất dữ liệu ra file csv mới, thiết lập index=False để không lưu thêm cột chỉ số dòng dư thừa
df.to_csv("processed_dulieuxettuyendaihoc.csv", index=False)

print(
    "Đã lưu file thành công với tên: processed_dulieuxettuyendaihoc.csv"
)
```
